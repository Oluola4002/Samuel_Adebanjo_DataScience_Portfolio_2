# Data Exploration and Visualization
This notebook demonstrates basic visualization techniques using **Matplotlib** and **Seaborn**, covering:
1. Line plots and scatter plots (Matplotlib)
2. Histograms and box plots (Seaborn)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a synthetic dataset for demonstration
np.random.seed(42)
n_samples = 100

dates = pd.date_range(start="2023-01-01", periods=n_samples, freq="D")
trend = np.linspace(10, 50, n_samples)
noise = np.random.normal(0, 5, n_samples)
values = trend + noise

category = np.random.choice(["Group A", "Group B", "Group C"], size=n_samples)
feature_x = np.random.uniform(10, 100, n_samples)
feature_y = 2.5 * feature_x + np.random.normal(0, 15, n_samples)

df = pd.DataFrame({
    "Date": dates,
    "Value": values,
    "Category": category,
    "Feature_X": feature_x,
    "Feature_Y": feature_y
})

display(df.head())
```

## 1. Matplotlib: Line Plot & Scatter Plot

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Line Plot - Trend over time
axes[0].plot(df["Date"], df["Value"], color="tab:blue", label="Value Trend")
axes[0].set_title("Line Plot: Value Trend Over Time")
axes[0].set_xlabel("Date")
axes[0].set_ylabel("Value")
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, linestyle="--", alpha=0.6)
axes[0].legend()

# Scatter Plot - Correlation between Feature X and Feature Y
axes[1].scatter(df["Feature_X"], df["Feature_Y"], color="tab:orange", alpha=0.8, edgecolors="k")
axes[1].set_title("Scatter Plot: Feature X vs Feature Y")
axes[1].set_xlabel("Feature X")
axes[1].set_ylabel("Feature Y")
axes[1].grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

## 2. Seaborn: Histogram & Box Plot

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram - Distribution of Value
sns.histplot(data=df, x="Value", kde=True, ax=axes[0], color="purple")
axes[0].set_title("Histogram: Distribution of Value")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")

# Box Plot - Value distribution by Category
sns.boxplot(data=df, x="Category", y="Value", ax=axes[1], hue="Category", palette="Set2", legend=False)
axes[1].set_title("Box Plot: Value by Category")
axes[1].set_xlabel("Category")
axes[1].set_ylabel("Value")

plt.tight_layout()
plt.show()
```

## 3. Seaborn: Correlation Heatmap
Heatmaps allow rapid scanning of linear associations across continuous features.

```python
# Correlation matrix for numerical attributes
corr = df[["Value", "Feature_X", "Feature_Y"]].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(corr, annot=True, cmap="Blues", fmt=".2f", square=True, linewidths=0.5)
plt.title("Correlation Heatmap: Numerical Features", fontsize=11, fontweight="bold")
plt.tight_layout()
plt.show()
```

## 4. Visualization Best Practices & Guidelines

| Goal | Recommended Chart | Library |
| :--- | :--- | :--- |
| **Trends over time** | Line Plot (`plt.plot` / `sns.lineplot`) | Matplotlib / Seaborn |
| **Feature Correlation** | Scatter Plot (`plt.scatter` / `sns.scatterplot`) | Matplotlib / Seaborn |
| **Distribution & Spread** | Histogram + KDE (`sns.histplot`) | Seaborn |
| **Category Comparison** | Box Plot / Violin Plot (`sns.boxplot`) | Seaborn |
| **Multi-variable Matrix** | Heatmap (`sns.heatmap`) | Seaborn |
